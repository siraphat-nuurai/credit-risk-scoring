```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
```

```python
# Load raw training data
df = pd.read_csv('../data/raw/credit-score-train.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Visualize the target variable (Class Imbalance Check)
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Credit_Score', order=['Good', 'Standard', 'Poor'], palette='viridis')
plt.title("Distribution of Credit Risk Tiers")
plt.show()
```

```python
# Bivariate Analysis: Debt vs Risk Tier
# We need to clean 'Outstanding_Debt' first just for the plot as it has string artifacts
df['Outstanding_Debt_Clean'] = pd.to_numeric(df['Outstanding_Debt'].str.replace('_', ''), errors='coerce')

plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Credit_Score', y='Outstanding_Debt_Clean', order=['Good', 'Standard', 'Poor'], palette='magma')
plt.title("Impact of Outstanding Debt on Risk Classification")
plt.ylim(0, 5000) # Capping for better visualization
plt.show()
```
